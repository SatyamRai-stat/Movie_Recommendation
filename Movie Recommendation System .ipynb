```python
import pandas as pd
import numpy as np
import warnings 
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('movies_metadata.csv')
df.head()
```

```python
df.columns,df.shape
```

```python
df.isnull().sum()
```

```python
df=df.drop_duplicates().reset_index(drop=True)
```

```python
df=df[['title','genres','overview','tagline','vote_average','popularity']]
```

```python
df.head()
```

```python
df=df.dropna(subset=['title'])
df['overview']=df['overview'].fillna(' ')
df['tagline']=df['tagline'].fillna(' ')
```

```python
df['vote_average']=df['vote_average'].fillna(df['vote_average'].mode()[0])
df['popularity']=df['popularity'].fillna(df['popularity'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['genres'][1]
```

```python
import ast
df['genres']=df['genres'].apply(lambda x:" ".join(i['name']  for i in ast.literal_eval(x)))
```

```python
df.head()
```

```python
df['tags']=df['genres']+" "+df['overview']+" "+df['tagline']
```

```python
df.head()
```

```python
import nltk 
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
from tqdm import tqdm
tqdm.pandas()
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_word=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
    
```

```python
def preprocess_text(text):
    text=text.lower()
    text=re.sub(r'[^a-zA-Z]',' ',text)
    words=text.split()
    words=[word for word in words if word not in stop_word]
    words=[lemmatizer.lemmatize(word) for word in words ]
    return " ".join(words)
```

```python
df['tags']=df['tags'].progress_apply(preprocess_text)
```

```python
df=df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index, index=df["title"]).drop_duplicates()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(tqdm(df['tags']))
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommned(title,n=10):
    if title not in indices:
        return ['Movie not found']
    idx=indices[title]
    sim_score=cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
    similar_idx=sim_score.argsort()[::-1][1:n+1]
    return df['title'].iloc[similar_idx]
```

```python
recommned('Jumanji')
```

```python
import pickle 
pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))
pickle.dump(indices,open('indices.pkl','wb'))
df.to_pickle('df.pkl')
pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
